```python
import math
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
while not ((ROOT / 'main.tex').exists() and (ROOT / 'refs.bib').exists()) and ROOT != ROOT.parent:
    ROOT = ROOT.parent
FIG_DIR = ROOT / 'figures' / 'hydrogen'
FIG_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    'figure.dpi': 120,
    'axes.titlesize': 12,
    'axes.labelsize': 10,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
})

a0 = 5.291772191461861e-11  # m
fe = 1.2355899446144601e20  # Hz
ell2 = math.log(3/2, 3)
rloss = (1 + 8*ell2)/24

rows = []
for n in range(1, 5):
    rows.append({
        'n': n,
        'radius_m': (n**2)*a0,
        'radius_Angstrom': (n**2)*a0*1e10,
    })
df = pd.DataFrame(rows)
df
```

Visible-note row anchors: shell band $3$, $B^*=137$, $\Lambda_b=6$, cadence witness $5/12$, and $R_{\mathrm{loss}}=(1+8\ell_2)/24$.


```python
anchor = pd.DataFrame([{
    'row': 'C4.A',
    'shell_band': 3,
    'B_star': 137,
    'Lambda_b': 6,
    'R_loss_trit_per_bip': rloss,
    'cadence_witness': '5/12',
    'rendering_note': 'local shell report',
    'f_e_Hz': fe,
}])
anchor
```

```python
fig, ax = plt.subplots(figsize=(6.0, 4.0))
ax.plot(df['n'], df['radius_Angstrom'], marker='o', linewidth=1.5)
ax.set_xlabel('n')
ax.set_ylabel('radius (Angstrom)')
ax.set_title('Hydrogen shell radii (render-layer)')
ax.grid(True, alpha=0.25)
out_path = FIG_DIR / 'hydrogen_shell_radii.png'
fig.tight_layout()
fig.savefig(out_path, dpi=180, bbox_inches='tight')
out_path.as_posix()
```

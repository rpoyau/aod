# C1 Solar 100-year worked anchors

This notebook mirrors the compact calculation spine used in the modular C1 section:

- input dictionary anchors
- body-to-rate bridge $m \to m_z$
- quarter-level count $\lambda = \log_3(m_z)/4$
- 100-year orbit-count renderings
- shell/ring figure artifacts
- derived source monon beat / causality-speed witness


```python
import math
import pandas as pd

c = 299792458.0
h = 6.62607015e-34

def m_z(m):
    return m*c*c/h

def lam(m):
    return math.log(m_z(m), 3)/4.0

masses = {
    'Sun': 1.98892e30,
    'Mercury': 3.3011e23,
    'Venus': 4.8675e24,
    'Earth': 5.97219e24,
    'Moon': 7.34767e22,
    'Mars': 6.4171e23,
    'Jupiter': 1.89819e27,
    'Saturn': 5.6834e26,
    'Uranus': 8.6810e25,
    'Neptune': 1.02413e26,
}

periods = {
    'Mercury': 0.2408467,
    'Venus': 0.61519726,
    'Earth': 1.0000174,
    'Mars': 1.8808476,
    'Jupiter': 11.862615,
    'Saturn': 29.447498,
    'Uranus': 84.016846,
    'Neptune': 164.79132,
}


ell2 = math.log(3/2, 3)
Rloss = (1 + 8*ell2)/24
Lambda_b = 6
loss_ticks = math.floor(24*Rloss)
chi4_source = lam(masses['Sun']) - Lambda_b - loss_ticks
source_cycles_100y = 100.0/chi4_source

rows = []
for body, m in masses.items():
    rows.append({
        'body': body,
        'mass_kg': m,
        'm_z_s^-1': m_z(m),
        'lambda_quarter': lam(m),
        'cycles_100y': (100.0/periods[body]) if body in periods else None,
        'source_monon_beat_y': chi4_source if body == 'Sun' else None,
        'source_monon_cycles_100y': source_cycles_100y if body == 'Sun' else None,
    })

df = pd.DataFrame(rows)
df
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
while not ((ROOT / 'main.tex').exists() and (ROOT / 'refs.bib').exists()) and ROOT != ROOT.parent:
    ROOT = ROOT.parent
FIG_DIR = ROOT / 'figures' / 'bacteria'
FIG_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    'figure.dpi': 120,
    'axes.titlesize': 12,
    'axes.labelsize': 10,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
})

np.random.seed(43)
steps = np.array([[0.35,0.25],[-0.2,0.6],[-0.6,0.25],[-0.35,-0.65],[0.55,-0.8],[0.8,-0.2],[0.45,0.65]])
pts = np.vstack([[0,0], np.cumsum(steps, axis=0)])
row_anchor = pd.DataFrame([{
    'row': 'C7.A',
    'structural_key': 'tetron:*:[bacteria]:6',
    'B_star': 2,
    'Lambda_b': 11,
    'R_loss_note': 'mass-scaled service burden',
    'cadence_noise': 'jitter band',
    'drift': 'recurrent',
}])
row_anchor
```

```python
fig, ax = plt.subplots(figsize=(6.0, 5.4))
for r in [0.8, 1.4, 2.0]:
    circ = plt.Circle((0,0), r, fill=False, ls='--', lw=1.0, color='gray')
    ax.add_patch(circ)
ax.add_patch(plt.Rectangle((-0.15,-0.08),0.30,0.16,facecolor='lightgray',edgecolor='black'))
ax.plot(pts[:,0], pts[:,1], marker='o', linewidth=1.5)
ax.text(0.16,0.1,'bacteria', fontsize=10)
ax.set_aspect('equal', adjustable='box')
ax.axis('off')
ax.set_title('Bacteria ominion jitter path', pad=10)
out_path = FIG_DIR / 'bacteria_ominion_jitter.png'
fig.tight_layout()
fig.savefig(out_path, dpi=180, bbox_inches='tight')
out_path.as_posix()
```

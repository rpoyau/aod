## 2. Example E4: Native Rest-Channel Worked Table

**Setup:** A shell-qualified motif window with declared admissibility counts $N_m = (n_1, n_2, n_3)$.
We calculate the native primitive $R_{rest}$ [trit/bip] and the derived internal rate $R_{rest}^{(biz)}$ [biz].

```python
# E4. NATIVE REST-CHANNEL CALCULATOR
# ==================================

def calc_rest_channel(t_eval, n1, n2, n3, b_star, lambda_b):
    """Calculates native loss and derived biz rate based on admissibility counts."""
    # Primitive trit/bip
    r_rest_trit = (n1 + n2 * l2) / t_eval
    
    # Gamma loss and derived biz
    gamma_loss = 3**r_rest_trit
    t_gamma = gamma_loss - 1.0
    
    # Attenuation accumulator A_B = (B* / B0)^(-Lambda_b)
    a_b = (b_star / B0)**(-lambda_b)
    r_rest_biz = (t_gamma / B0) * a_b
    
    return r_rest_trit, r_rest_biz

# Rows from Supplement A
e4_data = [
    {"Row": "E4.A", "T_eval": 12, "N_m": (2, 4, 6), "B*": 5, "Lambda_b": 4},
    {"Row": "E4.B", "T_eval": 24, "N_m": (1, 8, 15), "B*": 137, "Lambda_b": 6}
]

results_e4 = []
for d in e4_data:
    r_trit, r_biz = calc_rest_channel(d["T_eval"], d["N_m"][0], d["N_m"][1], d["N_m"][2], d["B*"], d["Lambda_b"])
    results_e4.append({
        "Row": d["Row"],
        "T_eval [bip]": d["T_eval"],
        "N_m": str(d["N_m"]),
        "B*": d["B*"],
        "Lambda_b": d["Lambda_b"],
        "R_rest [trit/bip]": f"{r_trit:.6f}",
        "R_rest (biz)": f"{r_biz:.6e}"
    })

df_e4 = pd.DataFrame(results_e4)
print("\n[Example E4] Native Rest-Channel Output:")
display(df_e4)
```

## 4. Example E6: Collision/Export Worked Table & Figure F3 Witness

**Setup:** A declared export window with two incoming wavelet legs ($a, b$). The interaction is governed by the balanced ternary gate $\oplus_3$. The fundamental physical requirement is that the cut residual must balance to zero: $\mathcal{R}_{cut} = S_{out} - (S_{in} + S_{prod}) = 0$.

If the ternary gate leaves an unbalanced debt, a **packet witness** is emitted to close the ledger.

```python
# E6. COLLISION/EXPORT TERNARY CALCULUS
# =====================================

def ternary_gate(a, b):
    """Balanced ternary gate: a [+] b in {-1, 0, 1}."""
    # Map {-1, 0, 1} to Z_3 {2, 0, 1}
    z3_map = {-1: 2, 0: 0, 1: 1}
    inv_z3_map = {2: -1, 0: 0, 1: 1}
    
    val = (z3_map[a] + z3_map[b]) % 3
    return inv_z3_map[val]

def eval_export_cut(a, b):
    """Evaluates the cut residual and required export packet."""
    gate_out = ternary_gate(a, b)
    
    # To enforce R_cut = 0, if gate_out is non-zero, we must emit a packet
    # to balance the local capacity.
    if gate_out == -1:
        push_pull = "push"
        q_eff = "+1/2"
        witness = "emitted packet"
    elif gate_out == 1:
        push_pull = "pull"
        q_eff = "-1/2"
        witness = "emitted packet"
    else:
        push_pull = "balanced"
        q_eff = "0"
        witness = "none"
        
    return gate_out, push_pull, q_eff, witness

e6_inputs = [
    ("E6.A", 1, -1),
    ("E6.B", 1, 1),
    ("E6.C", -1, -1)
]

results_e6 = []
for row, a, b in e6_inputs:
    gate_out, pp, q_eff, wit = eval_export_cut(a, b)
    results_e6.append({
        "Row": row,
        "Inputs (a, b)": f"({a:+d}, {b:+d})",
        "a [+] b": f"{gate_out:+d}" if gate_out != 0 else "0",
        "push/pull": pp,
        "q_eff": q_eff,
        "export witness": wit,
        "R_cut": "0"
    })

df_e6 = pd.DataFrame(results_e6)
print("\n[Example E6] Collision/Export Output:")
display(df_e6)

# Visualisation: Figure F3 Witness (Cut Diagram for E6.B Push)
fig, ax = plt.subplots(figsize=(6, 4))

# Draw the export window box
ax.add_patch(patches.Rectangle((0.2, 0.2), 0.6, 0.6, fill=False, edgecolor='black', linewidth=2))
ax.text(0.5, 0.85, 'declared export window', ha='center', fontsize=12)

# Draw the ternary gate
ax.add_patch(patches.Rectangle((0.4, 0.35), 0.2, 0.3, fill=True, color='lightgray', edgecolor='black'))
ax.text(0.5, 0.5, r'$\oplus_3$', ha='center', va='center', fontsize=14)

# Incoming legs
ax.annotate('', xy=(0.4, 0.55), xytext=(0.0, 0.65), arrowprops=dict(arrowstyle='->', lw=2))
ax.annotate('', xy=(0.4, 0.45), xytext=(0.0, 0.35), arrowprops=dict(arrowstyle='->', lw=2))
ax.text(0.1, 0.65, 'a', fontsize=12)
ax.text(0.1, 0.35, 'b', fontsize=12)
ax.text(0.2, 0.7, 'incoming legs', fontsize=11)

# Outgoing witness
ax.annotate('', xy=(0.95, 0.5), xytext=(0.6, 0.5), arrowprops=dict(arrowstyle='->', lw=2))
ax.text(0.75, 0.55, 'emitted packet witness', ha='center', fontsize=11)

ax.text(0.5, 0.1, r'$\mathcal{R}_{cut}=0$ on the same ledger', ha='center', fontsize=12)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')
plt.title("Figure F3: Collision / export-window diagram")
plt.show()
```

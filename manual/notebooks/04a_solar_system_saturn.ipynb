# Saturn ring and moon-shell notebook

This runnable example exports the ordered ring/moon shell family used in the visible note.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

ROOT = Path.cwd().resolve()
while not ((ROOT / 'main.tex').exists() and (ROOT / 'refs.bib').exists()) and ROOT != ROOT.parent:
    ROOT = ROOT.parent
FIG_DIR = ROOT / 'figures' / 'saturn'
FIG_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    'figure.dpi': 120,
    'axes.titlesize': 12,
    'axes.labelsize': 10,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
})

manifest = pd.DataFrame([
    {'name': 'D--E ring family', 'radius_witness': 'ordered inner shell set', 'period_witness': 'shell family', 'role': 'ring witness', 'note': 'multiple visible shell bands with Cassini Division gap'},
    {'name': 'Mimas', 'radius_witness': '185,539 km', 'period_witness': '0.942 d', 'role': 'outer shell witness', 'note': 'first selected moon shell beyond the main rings'},
    {'name': 'Enceladus', 'radius_witness': '237,948 km', 'period_witness': '1.370 d', 'role': 'outer shell witness', 'note': 'compact icy moon shell'},
    {'name': 'Tethys', 'radius_witness': '294,619 km', 'period_witness': '1.888 d', 'role': 'outer shell witness', 'note': 'mid-shell witness'},
    {'name': 'Dione', 'radius_witness': '377,396 km', 'period_witness': '2.737 d', 'role': 'outer shell witness', 'note': 'mid-outer shell witness'},
    {'name': 'Rhea', 'radius_witness': '527,069 km', 'period_witness': '4.518 d', 'role': 'outer shell witness', 'note': 'broad outer shell witness'},
    {'name': 'Titan', 'radius_witness': '1,221,870 km', 'period_witness': '15.945 d', 'role': 'slow outer shell', 'note': 'large far-shell witness'},
])
manifest
```

```python
fig, ax = plt.subplots(figsize=(8.0, 8.0))
ax.add_patch(Circle((0,0), 0.45, facecolor='gold', edgecolor='black'))
ax.text(0,0,'Saturn', ha='center', va='center', fontsize=11)
ring_rs = [0.8,1.1,1.45,1.75,2.1,2.45,2.8]
moon_rs = [3.35,3.95,4.6,5.25,6.05,8.2]
for r in ring_rs:
    ax.add_patch(Circle((0,0), r, fill=False, ls='--', lw=1.0, edgecolor='gray'))
ax.add_patch(Circle((0,0), 1.58, fill=False, lw=2.0, edgecolor='white'))
ax.add_patch(Circle((0,0), 1.58, fill=False, ls='--', lw=1.0, edgecolor='gray'))
for r in moon_rs:
    ax.add_patch(Circle((0,0), r, fill=False, ls='--', lw=1.0, edgecolor='lightgray'))
    ax.plot([r],[0],'o', ms=3, color='black')
ax.text(4.25,2.8,'rings: D, C, B, A, F, G, E', fontsize=10)
ax.text(4.55,-2.8,'moons: Mimas, Enceladus\nTethys, Dione, Rhea, Titan', fontsize=10)
ax.set_aspect('equal', adjustable='box')
ax.set_xlim(-9.0, 9.8)
ax.set_ylim(-9.5, 9.3)
ax.axis('off')
ax.set_title('Saturn multi-shell witness', pad=12)
out_path = FIG_DIR / 'saturn_shell_family.png'
fig.tight_layout()
fig.savefig(out_path, dpi=180, bbox_inches='tight')
out_path.as_posix()
```

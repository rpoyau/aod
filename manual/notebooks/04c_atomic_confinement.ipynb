# Nucleus confinement witness notebook

This notebook reproduces the compact short-window confinement witness table used in the somatic-mechanics chapter.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
while not ((ROOT / 'main.tex').exists() and (ROOT / 'refs.bib').exists()) and ROOT != ROOT.parent:
    ROOT = ROOT.parent
FIG_DIR = ROOT / 'figures' / 'nucleus'
FIG_DIR.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    'figure.dpi': 120,
    'axes.titlesize': 12,
    'axes.labelsize': 10,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
})

df = pd.DataFrame([
    {
        'row': 'C5.A',
        'structural_key': 'tetron:3:3:6',
        'signature': '(lambda=3/2,137,rho2,V4)',
        'B_star': 137,
        'Lambda_b': 6,
        'proximity': 'close',
        'confinement': 'confined',
        'transmutation': 0,
    },
    {
        'row': 'C5.B',
        'structural_key': 'tetron:3:3:6',
        'signature': '(lambda=3/2,139,rho3,V4)',
        'B_star': 139,
        'Lambda_b': 3,
        'proximity': 'failed',
        'confinement': 'unconfined',
        'transmutation': 1,
    },
])
df
```

```python
fig, ax = plt.subplots(figsize=(6.2, 4.0))
ax.bar(df['row'], df['B_star'])
for i, v in enumerate(df['B_star']):
    ax.text(i, v + 1.5, str(v), ha='center', fontsize=9)
ax.set_ylabel('B*')
ax.set_title('Short-window confinement witness')
ax.grid(axis='y', alpha=0.25)
out_path = FIG_DIR / 'nucleus_confinement_witness.png'
fig.tight_layout()
fig.savefig(out_path, dpi=180, bbox_inches='tight')
out_path.as_posix()
```
